## Cell 3 — Verify runtime and source data; choose Qwen or explicit NLI fallback

```python
import sys, json, hashlib, subprocess
import torch
sys.path.insert(0, str(SOURCE))
from real_pipeline.common import sha256, read_rows, validate_posts, require_disjoint, load_sealed_export
from real_pipeline.language import language_histogram, language_scope
from real_pipeline.contracts import STUDENT_CONTRACT, model_identity
assert torch.cuda.is_available(), 'A separate Colab GPU runtime is required'
print({'gpu': torch.cuda.get_device_name(0), 'memory_bytes': torch.cuda.get_device_properties(0).total_memory})
BEHAVIOR_TEACHER = 'qwen'  # Explicit fallback: 'nli'. Never silently substitute.
assert BEHAVIOR_TEACHER in ('qwen', 'nli')
if BEHAVIOR_TEACHER == 'qwen':
    assert torch.cuda.get_device_properties(0).total_memory >= 38_000_000_000, 'Choose A100/H100 or explicit nli fallback'
BANK = LAKE / 'training-banks/phase3_health_training_150k_v1_unlabeled.parquet'
HOLDOUT = LAKE / 'eval/phase3_holdout_600_unlabeled.jsonl'
EXPORT_MANIFEST = LAKE / 'eval/phase3_v1_export_manifest.json'
PAIRS = LAKE / 'training-banks/phase5_social_encoder_pairs_v1.parquet'
RUN = LAKE / f'runs/real-v1.2-{BEHAVIOR_TEACHER}'
RUN.mkdir(parents=True, exist_ok=True)
source_manifest = json.loads((SOURCE / 'source_manifest.json').read_text())
for relative, digest in source_manifest['files'].items():
    assert sha256(SOURCE / relative) == digest, f'Implementation source changed: {relative}'
print({'implementation_source_manifest': sha256(SOURCE / 'source_manifest.json')})
seal, holdout_rows = load_sealed_export(HOLDOUT, EXPORT_MANIFEST)
assert sha256(BANK) == seal['bank']['sha256'], 'Bank digest differs from the audited seal'
train_rows = read_rows(BANK)
validate_posts(train_rows); require_disjoint(train_rows, holdout_rows)
print({'bank_rows': len(train_rows), 'holdout_rows': len(holdout_rows), 'bank_sha256': sha256(BANK), 'holdout_sha256': sha256(HOLDOUT), 'outputs': str(RUN)})
# Subprocesses release all GPU allocations between stages. Failures stop the cell.
def run(*args):
    subprocess.run([sys.executable, '-u', str(SOURCE / 'real_pipeline.py'), *map(str,args)], check=True)

print({'language_audit': language_histogram(train_rows), 'student_contract': STUDENT_CONTRACT, 'behavior_model': model_identity(BEHAVIOR_TEACHER)})
OBJECTIVE = 'health'  # 'joint' additionally requires fully scoped v1 triplets.
```

## Cell 4 — Label real training posts and the held-out review set
Durable shards resume with unchanged input/model/rubric identity. Qwen uses single-token 0–9 log probabilities, not generated prose. NLI fallback identity is retained. Rows outside declared-English scope remain present with null scores and an explicit unscored reason. No human quality metric is computed.

```python
for name, source, split in [('train', BANK, 'train'), ('holdout', HOLDOUT, 'holdout')]:
    common = ['--input', source, '--holdout', HOLDOUT, '--export-manifest', EXPORT_MANIFEST, '--split', split, '--mode', 'colab', '--device', 'cuda']
    run('label', *common, '--teacher', 'political', '--max-length', 512, '--batch-size', 8, '--output', RUN / f'{name}-political')
    run('label', *common, '--teacher', BEHAVIOR_TEACHER, '--max-length', 1024 if BEHAVIOR_TEACHER == 'qwen' else 512, '--batch-size', 8, '--output', RUN / f'{name}-{BEHAVIOR_TEACHER}')
    run('merge', *common, '--political', RUN / f'{name}-political', '--behavior', RUN / f'{name}-{BEHAVIOR_TEACHER}', '--behavior-teacher', BEHAVIOR_TEACHER, '--output', RUN / f'{name}-labeled.parquet')
```

## Cell 6 — Validate pairs only for an explicit joint objective
A missing bank is an honest stop. Do not substitute the invalidated v0 bank. These are weak graph-membership positives, not verified co-engagement.

```python
if OBJECTIVE == 'joint':
    from real_pipeline.common import validate_pairs
    assert PAIRS.exists(), f'v1 graph rebuild/pair export still required: {PAIRS}'
    pairs, pair_manifest = validate_pairs(PAIRS, HOLDOUT, EXPORT_MANIFEST)
    print({'pairs': len(pairs), 'graph_version': pair_manifest['graph_version'], 'known_for_sha256': pair_manifest['known_for_sha256'], 'pair_sha256': sha256(PAIRS)})
    scoped_pairs = [r for r in pairs if all(language_scope(r, p)['in_scope'] for p in ('anchor', 'positive', 'negative'))]
    print({'in_scope_pairs': len(scoped_pairs), 'unscored_pairs': len(pairs)-len(scoped_pairs)})
    assert scoped_pairs, 'Joint training requires declared-English metadata on every triplet role'
else:
    print('Health-only training: social head remains untrained; no graph pair bank consumed.')
```

## Cell 7 — Train five components and export numerical candidates
5,000 steps is a provisional schedule, not a quality guarantee. Checkpoints persist every 250 steps. To resume, set RESUME to the latest checkpoint path and leave all settings unchanged. Integrity-checked recovery falls back to an older intact checkpoint, archives incomplete files and any later step-log records, and writes a recovery receipt. Checkpoints without integrity sidecars from the original v1 smoke are not silently trusted.
Revision 3 predicts politicalness, outrage, criticism, engagement bait and spam separately. Heat is not a trained target; serving aggregation requires a frozen definition and calibration. Four-output historical checkpoints cannot resume here. Hosted Drive and real optimizer interruption recovery remain untested.


```python
RESUME = None  # e.g. RUN / 'student/checkpoint-0000250.pt'
extra = ['--resume', RESUME] if RESUME else []
if OBJECTIVE == 'joint':
    extra += ['--pairs', PAIRS]
run('train', '--input', RUN / 'train-labeled.parquet', '--holdout', HOLDOUT, '--export-manifest', EXPORT_MANIFEST, '--objective', OBJECTIVE, '--mode', 'colab', '--device', 'cuda', '--steps', 5000, '--batch-size', 32, '--save-every', 250, '--max-length', 256, '--output', RUN / 'student', *extra)
```
